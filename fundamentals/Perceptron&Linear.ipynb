```python
from torchvision import datasets
import torchvision
train_data = datasets.MNIST(
    root='./data',
    train=True,
    download=True,
    transform=torchvision.transforms.ToTensor(),
    target_transform=None
)
test_data = datasets.MNIST(
    root="data",
    train=False,
    download=True,
    transform=torchvision.transforms.ToTensor(),
    target_transform=None
)
from torch.utils.data import DataLoader
BATCH_SIZE = 32
train_dataloader = DataLoader(dataset=train_data, batch_size=BATCH_SIZE, shuffle=True)
test_dataloader = DataLoader(dataset=test_data, batch_size=BATCH_SIZE, shuffle=False)
```

```python
import torch.nn.functional as F
class Perceptron(nn.Module):
    def __init__(self, in_features):
        super(Perceptron, self).__init__()
        self.weights = nn.Parameter(data=torch.rand(in_features, 1))
        self.bias = nn.Parameter(data=torch.rand(1,1))
    def forward(self,x):
        return x @ self.weights + self.bias
class ZachLinear(nn.Module):
    def __init__(self,in_features, out_features):
        super(ZachLinear, self).__init__()
        self.weights = nn.Parameter(data=torch.rand(in_features, out_features))
        self.bias = nn.Parameter(data=torch.rand(out_features))
    def forward(self,x):
        x = torch.matmul(x,self.weights)
        wbias = torch.add(x,self.bias)
        # print("X:",wbias, wbias.shape)
        return wbias
class blobsClassifier(nn.Module):
    def __init__(self):
        super(blobsClassifier, self).__init__()
        self.model = nn.Sequential(
            ZachLinear(in_features=2,out_features=10),
            nn.ReLU(),
            ZachLinear(in_features=10, out_features=10),
            nn.ReLU(),
            ZachLinear(in_features=10, out_features=3),
        )
    def forward(self, x):
        return self.model(x)
blobsClassification = blobsClassifier().cpu()
```

```python
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
X, y = make_blobs(n_samples=20, centers=3, n_features=2,
                  random_state=0)
plt.scatter(X[:, 0], X[:, 1], c=y)
```

```python
from tqdm import tqdm
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(params=blobsClassification.parameters(), lr=.01)
epochs = 1000
X = torch.Tensor(X).cpu()
y = torch.Tensor(y).to(torch.long).cpu()
for epoch in tqdm(range(epochs)):
    blobsClassification.train()
    y_pred = blobsClassification(X)
    loss = criterion(y_pred.squeeze(), y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
print(loss)
```

```python

def plot_decision_boundary(model: torch.nn.Module, X: torch.Tensor, y: torch.Tensor):
    """Plots decision boundaries of model predicting on X in comparison to y.

    Source - https://madewithml.com/courses/foundations/neural-networks/ (with modifications)
    """
    # Put everything to CPU (works better with NumPy + Matplotlib)
    model.to("cpu")
    X, y = X.to("cpu"), y.to("cpu")

    # Setup prediction boundaries and grid
    x_min, x_max = X[:, 0].min() - 0.1, X[:, 0].max() + 0.1
    y_min, y_max = X[:, 1].min() - 0.1, X[:, 1].max() + 0.1
    xx, yy = np.meshgrid(np.linspace(x_min, x_max, 101), np.linspace(y_min, y_max, 101))

    # Make features
    X_to_pred_on = torch.from_numpy(np.column_stack((xx.ravel(), yy.ravel()))).float()

    # Make predictions
    model.eval()
    with torch.inference_mode():
        y_logits = model(X_to_pred_on)

    # Test for multi-class or binary and adjust logits to prediction labels
    if len(torch.unique(y)) > 2:
        y_pred = torch.softmax(y_logits, dim=1).argmax(dim=1)  # mutli-class
    else:
        y_pred = torch.round(torch.sigmoid(y_logits))  # binary

    # Reshape preds and plot
    print(y_pred.shape, xx.shape)
    y_pred = y_pred.reshape(xx.shape).detach().numpy()
    plt.contourf(xx, yy, y_pred, cmap=plt.cm.RdYlBu, alpha=0.7)
    plt.scatter(X[:, 0], X[:, 1], c=y, s=40, cmap=plt.cm.RdYlBu)
    plt.xlim(xx.min(), xx.max())
    plt.ylim(yy.min(), yy.max())
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.title("Train")
plot_decision_boundary(blobsClassification, X, y)
```
